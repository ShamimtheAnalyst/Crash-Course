## **১. ভূমিকা (Introduction)**

### ১. Chi-Squared টেস্ট আসলে কী?

সহজ কথায়, **Chi-Squared টেস্ট** হলো "প্রত্যাশা বনাম বাস্তবতা" (Expectation vs. Reality)-র একটি লড়াই। যখন আমাদের কাছে ক্যাটেগরিক্যাল ডেটা (যেমন: লিঙ্গ, রঙ, প্রজাতি, বা সাইজ) থাকে, তখন আমরা এই টেস্ট ব্যবহার করি এটা দেখতে যে—আমাদের দেখা ডেটা কি কাকতালীয়, নাকি এর পেছনে কোনো নির্দিষ্ট কারণ বা সম্পর্ক আছে।

এটি মূলত **ফ্রিকোয়েন্সি (Frequency)** বা গণনার ওপর ভিত্তি করে কাজ করে, গড় (Mean) বা মিডিয়ানের ওপর নয়।

---

### ২. গাণিতিক সূত্র (The Formula)

টেস্টটি নিচের সূত্রের মাধ্যমে একটি স্কোর বের করে:

এখানে:

* **(Observed):** আপনি বাস্তবে যা দেখেছেন (আপনার ডেটাসেট)।
* **(Expected):** আপনি যা আশা করেছিলেন (যদি কোনো বিশেষ সম্পর্ক না থাকতো)।

---

### ৩. Chi-Squared টেস্টের প্রকারভেদ

#### ক. Goodness of Fit (একক ভ্যারিয়েবল)

এটি পরীক্ষা করে যে, একটি ভ্যারিয়েবলের ক্যাটেগরিগুলো কি একটি নির্দিষ্ট ডিস্ট্রিবিউশন মেনে চলে?

* **উদাহরণ:** ধরুন আপনি একটি ডাইস (ছক্কা) ৬০০ বার নিক্ষেপ করলেন। আপনি আশা করবেন প্রতিটি সংখ্যা (১-৬) ১০০ বার করে আসবে। কিন্তু বাস্তবে যদি দেখেন '৬' এসেছে ২০০ বার, তবে এই টেস্ট আপনাকে বলবে ছক্কাটি কি ত্রুটিপূর্ণ (Biased) নাকি স্বাভাবিক।

**Goodness of Fit:** একটি ভ্যারিয়েবলের বিভিন্ন ক্যাটেগরির অনুপাত কি আমাদের প্রত্যাশার সমান?

#### খ. Test of Independence (দুটি ভ্যারিয়েবল)

এটি পরীক্ষা করে যে, দুটি ক্যাটেগরিক্যাল ভ্যারিয়েবল কি একে অপরের সাথে জড়িত?

* **উদাহরণ:** আপনার দেওয়া `iris` ডেটাসেটে আমরা দেখছি "প্রজাতি" (Species) কি "ফুলের সাইজ" (Size)-এর ওপর প্রভাব ফেলে? যদি প্রভাব না ফেলত, তবে ছোট-বড় সব সাইজের ফুল সব প্রজাতির মধ্যে সমানভাবে পাওয়া যেত।

**Test of Independence:** দুটি ক্যাটেগরিক্যাল ভ্যারিয়েবলের মধ্যে কি কোনো সম্পর্ক আছে?

---

### ৪. টেস্টটি কীভাবে সিদ্ধান্ত নেয়?

Chi-Squared টেস্ট করার সময় আমরা দুটি কাল্পনিক জগত চিন্তা করি:
১. **Null Hypothesis ():** এখানে সবকিছু স্বাধীন এবং কোনো সম্পর্ক নেই। যা পার্থক্য দেখছেন তা কেবল by-Chance।
২. **Alternative Hypothesis ():** এখানে ভ্যারিয়েবলগুলোর মধ্যে বাস্তব সম্পর্ক আছে।

আমরা যখন **p-value** বের করি, তখন:

* **:** এর মানে হলো, "ভাগ্যের জোরে" এমন ডেটা পাওয়ার সম্ভাবনা ৫% এরও কম। তাই আমরা  বাতিল করি এবং বলি—হ্যাঁ, এখানে একটি **সম্পর্ক আছে**।
* **:** এর মানে হলো, পার্থক্যটি সামান্য এবং এটি কাকতালীয় হতে পারে। তাই আমরা বলি—কোনো **সম্পর্ক নেই**।

---

### ৫. Chi-Squared টেস্টের শর্তাবলি (Assumptions)

এই টেস্টটি করার আগে কিছু জিনিস নিশ্চিত করতে হয়, নাহলে রেজাল্ট ভুল আসতে পারে:

* **Expected Frequency:** প্রতিটি ঘরে প্রত্যাশিত মান (Expected value) কমপক্ষে ৫ হতে হবে। আপনার কোডে `test_independence$expected` দিয়ে আপনি ঠিক এই কাজটাই করেছেন।
* **Independence:** একটি পর্যবেক্ষণ অন্যটির ওপর নির্ভর করবে না (একই ফুলের তথ্য দুবার নেওয়া যাবে না)।
* **Random Sampling:** ডেটা লটারি বা র্যান্ডম উপায়ে সংগ্রহ করতে হবে।

---

### ৬. প্রো-টিপ: কখন Fisher's Exact Test ব্যবহার করবেন?

যদি আপনার ডেটাসেট খুব ছোট হয় এবং কোনো একটি ঘরে (Cell) **Expected Value ৫-এর কম** চলে আসে, তবে Chi-Squared টেস্ট আর সঠিক ফল দেয় না। তখন আপনাকে **Fisher's Exact Test** ব্যবহার করতে হবে। R-এ এটি করা খুব সহজ: `fisher.test(table(flower))`.

## **২. এনভায়রনমেন্ট সেটআপ (Environment Setup)**

প্রথমে প্রয়োজনীয় লাইব্রেরি লোড করি এবং `iris` ডাটাসেটের 'Sepal.Length' কে ক্যাটেগরিক্যাল ভ্যারিয়েবলে রূপান্তর করি।

```r
library(tidyverse)
```

```r
# Data preparation
flower <- iris %>%
  mutate(size = cut(Sepal.Length,
                    breaks = 3,
                    labels = c("Small", "Medium", "Large"))) %>%
  select(Species, size)
```

```r
# টেবিল ফরম্যাটে ডেটা দেখা
table(flower)
```

## **৩. Chi-Squared Goodness of Fit Test**

এখানে আমরা পরীক্ষা করছি: **ফুলের সাইজ (Small, Medium, Large) কি সমান অনুপাতে (Equal Proportions) আছে?**

* **:** ফুলের সাইজের অনুপাতগুলো সমান।
* **:** ফুলের সাইজের অনুপাতগুলো সমান নয়।

```r
# ভিজ্যুয়ালি দেখা
flower %>%
  ggplot(aes(x = size)) +
  geom_bar(fill = "#a139ca", alpha = 0.7) +
  labs(title = "Goodness of Fit Test",
       subtitle = "Proportion of flowers by size") +
  theme_test(base_size = 15)
```

```r
# টেস্ট সম্পাদন
goodness_fit <- flower %>% select(size) %>% table() %>% chisq.test()
goodness_fit
```

**ইন্টারপ্রিটেশন (Interpretation):**
এখানে -value = , যা  এর চেয়ে অনেক কম।

> **সিদ্ধান্ত:** আমরা Null Hypothesis () রিজেক্ট করছি। এর মানে হলো ফুলের সাইজগুলো সমান অনুপাতে নেই। কোনো নির্দিষ্ট সাইজের ফুল অন্যগুলোর চেয়ে উল্লেখযোগ্যভাবে বেশি বা কম আছে।

## **৪. Chi-Squared Test of Independence**

এখানে প্রশ্ন হলো: **ফুলের প্রজাতি (Species) এবং সাইজের (Size) মধ্যে কি কোনো সম্পর্ক আছে?**

* **:** প্রজাতি এবং সাইজ একে অপরের থেকে স্বাধীন (Independent)।
* **:** প্রজাতি এবং সাইজ একে অপরের ওপর নির্ভরশীল (Dependent)।

```r
# স্ট্যাকড বার চার্ট (Visualizing Independence)
flower %>%
  ggplot(aes(size, fill = Species)) +
  geom_bar(alpha = 0.7) +
  labs(title = "Chi Squared Test of Independence",
       subtitle = "Iris by Species and Size") +
  theme_test(base_size = 15) +
  scale_fill_manual(values = c("setosa" = "#d48ce1", "versicolor" = "#a139ca", "virginica" = "#6d2683"))
```

```r
# টেস্ট সম্পাদন
test_independence <- table(flower) %>% chisq.test()
test_independence
```

**ইন্টারপ্রিটেশন (Interpretation):**
এখানে -value = । যেহেতু , তাই আমরা বলতে পারি **প্রজাতি এবং সাইজ একে অপরের ওপর নির্ভরশীল।** > **সহজ ভাষায়:** একটি ফুল কোন প্রজাতির, তা জানলে সেই ফুলের সাইজ কেমন হতে পারে তা আমরা প্রেডিক্ট করতে পারব (যেমন: Setosa ফুলগুলো সাধারণত ছোট হয়)।

---

## **৫. এক্সপেক্টেড ভ্যালু চেক (Expected Values Check)**

Chi-Squared টেস্টের একটি গুরুত্বপূর্ণ শর্ত হলো প্রতিটি সেলের **Expected Frequency** কমপক্ষে ৫ হতে হবে।

```r
# Expected values দেখা
test_independence$expected
```

যদি কোনো সেলে Expected Value ৫ এর কম হয়, তবে **Fisher's Exact Test** ব্যবহার করা বেশি নির্ভরযোগ্য।

## **সারসংক্ষেপ (Key Takeaways)**

1. **Expected Frequency:** এটি সেই মান যা আমরা আশা করি যদি ভ্যারিয়েবলগুলো পুরোপুরি স্বাধীন হতো।
2. **Observed Frequency:** এটি আমাদের ডাটাসেটের বাস্তব ডেটা।
3. **p-value Interpretation:**  মানে হলো আমাদের পর্যবেক্ষণে প্রাপ্ত পার্থক্যটি স্ট্যাটিস্টিক্যালি তাৎপর্যপূর্ণ (Significantly different)।
