### থিম (Themes) কী?

```r
library(tidyverse)
```

```r
# একটি বেসিক স্ক্যাটার প্লট (আমাদের বেস প্লট 'p')
p <- ggplot(mtcars, aes(wt, mpg)) +
  geom_point() +
  ggtitle("Weight vs MPG") +
  xlab("Weight (1000 lbs)") +
  ylab("Miles per Gallon")
```

### ২. প্রিসেট থিম ব্যবহার (Using Preset Themes)
ggplot2 এ বেশ কিছু তৈরি করা (Preset) থিম থাকে যা এক লাইনেই আপনার প্লটের চেহারা বদলে দিতে পারে।

### ৩. নিজের মতো করে থিম সাজানো (Customizing Themes)
- theme() ফাংশনের ভেতরে তিনটি প্রধান এলিমেন্ট ফাংশন কাজ করে:
- element_text(): টেক্সট বা লেখার স্টাইল পরিবর্তন করতে।
- element_line(): লাইন বা গ্রিড পরিবর্তন করতে।
- element_rect(): ব্যাকগ্রাউন্ডের আয়তাকার এরিয়া পরিবর্তন করতে।

#### ক) শিরোনাম এবং অক্ষ (Titles & Axes) কাস্টমাইজেশন

```r
p + theme(
    plot.title = element_text(color = "steelblue", family = "Arial", face = "bold", size = rel(2), hjust = 0.5),
    axis.title.x = element_text(color = "red", face = "bold.italic"),
    axis.line.x = element_line(linewidth = 2, color = "steelblue", linetype = "dotted")
  )
```

**নোট:** size এর বদলে এখন linewidth ব্যবহার করা হয় (Warning অনুযায়ী আপডেট করা হয়েছে)।

#### খ) প্যানেল এবং ব্যাকগ্রাউন্ড পরিবর্তন

```r
p + theme(
    panel.background = element_rect(fill = "lightyellow", color = "lightblue", linewidth = 2),
    panel.grid.major.x = element_line(linewidth = 1, color = "grey80")
  )
```

### ৪. লেজেন্ড কাস্টমাইজেশন (Legend Customization)
লেজেন্ড বা সূচিপত্রকে সুন্দরভাবে সাজালে গ্রাফটি পড়তে সুবিধা হয়।

```r
p_legend <- ggplot(mtcars, aes(wt, mpg, color = factor(cyl))) +
  geom_point() +
  labs(color = "Cylinders")

p_legend + theme(
    legend.position = "bottom", # লেজেন্ড নিচে নেওয়া
    legend.background = element_rect(fill = "snow", color = "black"),
    legend.key = element_rect(fill = "white")
  )
```

### ৫. ফেসট প্লট কাস্টমাইজেশন (Facet Themes)
যখন আমরা facet_wrap() ব্যবহার করি, তখন উপরের স্ট্রিপ বা প্যানেলের শিরোনামও কাস্টমাইজ করা যায়।

```r
p_facet <- ggplot(mtcars, aes(wt, mpg)) + geom_point() + facet_wrap(~cyl)

p_facet + theme(
    strip.background = element_rect(fill = "lightblue", color = "steelblue", linewidth = 2),
    strip.text = element_text(size = 12, face = "bold", color = "darkblue")
  )
```

### ৬. গ্লোবাল থিম সেটআপ (Setting the Default Theme)
আপনি যদি চান আপনার পুরো প্রজেক্টের সব গ্রাফ একই থিমে হবে, তবে শুরুর দিকেই theme_set() করে নিন।

```r
theme_set(
  theme_minimal() +
  theme(plot.title = element_text(color = "steelblue", face = "bold", hjust = 0.5))
)

# এখন সব প্লট অটোমেটিক্যালি এই স্টাইল ফলো করবে!
```

### এক নজরে আপনার শেখা টুলকিট (Cheat Sheet):

- theme_minimal(): দ্রুত সুন্দর গ্রাফের জন্য।
- element_text(family, face, color): লেখা সুন্দর করার জন্য।
- legend.position: লেজেন্ডের জায়গা পরিবর্তনের জন্য (top, bottom, left, right, inside)।
- linewidth: লাইনের পুরুত্ব পরিবর্তনের জন্য (পুরানো size এর বদলে)।
- axis.text.x = element_text(angle = 45): যদি এক্সিসের নামগুলো গায়ে গায়ে লেগে যায় তবে সেগুলোকে ঘুরিয়ে দেওয়ার জন্য।
