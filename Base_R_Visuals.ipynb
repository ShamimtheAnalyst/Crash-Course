# Base R Visuals Step-by-Step Guide
Base R-এ প্লটিং হলো অ্যাডিটিভ প্রক্রিয়া। আপনি একটি বেসিক কমান্ড দিয়ে শুরু করেন এবং তারপর অতিরিক্ত ফাংশন ব্যবহার করে "লেয়ার" যোগ করতে থাকেন।

## পর্ব ১: ফাউন্ডেশন তৈরি (plot() ফাংশন)
প্রতিটি ভিজ্যুয়ালাইজেশন শুরু হয় plot() ফাংশন দিয়ে। এটি অ্যাক্সিস এবং ক্যানভাস তৈরি করে।

## ধাপ ১.১: বেসিক ক্যানভাস

```r
# ধাপ ১: শুধুমাত্র ডাটা
plot(mtcars$wt, mtcars$mpg)
```

### ব্যাখ্যা:
* plot(x, y) ফাংশন দুটি ভেরিয়েবলের মধ্যে সম্পর্ক দেখায়
* mtcars$wt: এক্স-অ্যাক্সিস (ওজন)

* mtcars$mpg: ওয়াই-অ্যাক্সিস (মাইলেজ)
* এটি সবচেয়ে সহজ স্ক্যাটার প্লট

## ধাপ ১.২: লেবেল এবং টাইটেল যোগ

```r
# ধাপ ২: মেইন টাইটেল এবং অ্যাক্সিস লেবেল যোগ
plot(mtcars$wt, mtcars$mpg,
     main = "গাড়ির ওজন বনাম জ্বালানি দক্ষতা",
     xlab = "ওজন (1000 পাউন্ড)",
     ylab = "মাইল প্রতি গ্যালন (MPG)")
```

ব্যাখ্যা:
* main: গ্রাফের প্রধান শিরোনাম
* xlab: এক্স-অ্যাক্সিসের লেবেল
* ylab: ওয়াই-অ্যাক্সিসের লেবেল
* প্রফেশনাল টিপ: লেবেল স্পষ্ট রাখুন যাতে অন্যরাও গ্রাফ বুঝতে পারে

## ধাপ ১.৩: স্টাইলিং (pch এবং col)

```r
# ধাপ ৩: পয়েন্ট স্টাইল এবং কালার পরিবর্তন
plot(mtcars$wt, mtcars$mpg,
     main = "স্টাইল করা স্ক্যাটার প্লট",
     xlab = "ওজন",
     ylab = "MPG",
     pch = 16,          # 16 হলো সলিড সার্কেল
     col = "steelblue", # পয়েন্টের রং
     cex = 1.2)         # পয়েন্ট সাইজ (1.2 গুণ বড়)
```

pch ভ্যালুর রেফারেন্স:
* 1: ○ (খালি বৃত্ত)
* 16: ● (সলিড বৃত্ত)
* 17: ▲ (ত্রিভুজ)
* 18: ◆ (রম্বস)
* 19: ● (বড় সলিড বৃত্ত)

# পর্ব ২: স্ট্রাকচারাল লেয়ার যোগ (abline এবং grid)
ক্যানভাস তৈরি হলে, আমরা ডাটা ইন্টারপ্রেট করতে সাহায্যকারী লেয়ার যোগ করি।

## ধাপ ২.১: ট্রেন্ড লাইন যোগ

```r
# প্রথমে বেসিক প্লট
plot(mtcars$wt, mtcars$mpg,
     main = "ট্রেন্ড লাইন সহ স্ক্যাটার প্লট",
     xlab = "ওজন",
     ylab = "MPG",
     pch = 16,
     col = "darkgray")
```

```r
# প্রথমে বেসিক প্লট
plot(mtcars$wt, mtcars$mpg,
     main = "ট্রেন্ড লাইন সহ স্ক্যাটার প্লট",
     xlab = "ওজন",
     ylab = "MPG",
     pch = 16,
     col = "darkgray")

# লাল রঙের রিগ্রেশন লাইন যোগ
abline(lm(mpg ~ wt, data = mtcars),
       col = "red",
       lwd = 2,      # লাইনের পুরুত্ব
       lty = 1)      # লাইন টাইপ (1 = সলিড)
```

### ব্যাখ্যা:
* abline(): সরল রেখা যোগ করে
* lm(mpg ~ wt): লিনিয়ার মডেল তৈরি করে (MPG ওজনের উপর নির্ভর করে)
* lwd: Line Width (ডিফল্ট 1)

## ধাপ ২.২: গ্রিড লাইন যোগ

```r
plot(mtcars$wt, mtcars$mpg,
     pch = 16,
     col = "purple",
     main = "গ্রিড সহ প্লট")
```

### গ্রিড টাইপস:

* "dashed": ড্যাশড লাইন
* "dotted": ডটেড লাইন
* "dotdash": ডট-ড্যাশ লাইন

```r
cylinder_counts <- table(mtcars$cyl)

# বারপ্লট তৈরি
barplot(cylinder_counts,
        main = "সিলিন্ডার সংখ্যা অনুযায়ী গাড়ি",
        xlab = "সিলিন্ডার সংখ্যা",
        ylab = "গাড়ির সংখ্যা",
        col = rainbow(3),          # রংবেরঙের বার
        border = "black",
        ylim = c(0, 15))          # ওয়াই-অ্যাক্সিস লিমিট


# বারগুলোর উপরে সংখ্যা লেখা
text(1:3,                            # এক্স পজিশন
     cylinder_counts + 0.5,          # ওয়াই পজিশন (বারের উপরে)
     labels = cylinder_counts,       # লেবেল
     col = "black",                  # রং
     font = 2)                       # ফন্ট স্টাইল (2 = বোল্ড)
```

```r
# ডেনসিটি প্লট তৈরি
dens <- density(mtcars$mpg)

plot(dens,
     main = "MPG-এর ডেনসিটি ডিস্ট্রিবিউশন",
     xlab = "MPG",
     ylab = "ঘনত্ব",
     lwd = 2,
     col = "navy")

# এরিয়া ফিল করা
polygon(dens,
        col = adjustcolor("lightblue", alpha.f = 0.5),  # স্বচ্ছ রং
        border = "navy")

# মিডিয়ান লাইন
abline(v = median(mtcars$mpg),
       col = "red",
       lty = 2,
       lwd = 1.5)
```

```r
# ডেনসিটি প্লট তৈরি
dens <- density(mtcars$mpg)

plot(dens,
     main = "MPG-এর ডেনসিটি ডিস্ট্রিবিউশন",
     xlab = "MPG",
     ylab = "ঘনত্ব",
     lwd = 2,
     col = "navy")

# এরিয়া ফিল করা
polygon(dens,
        col = adjustcolor("lightblue", alpha.f = 0.5),  # স্বচ্ছ রং
        border = "navy")

# মিডিয়ান লাইন
abline(v = median(mtcars$mpg),
       col = "red",
       lty = 2,
       lwd = 1.5)
# লেজেন্ড
legend("topright",
       legend = c("ডেনসিটি", "মিডিয়ান"),
       col = c("navy", "red"),
       lty = c(1, 2),
       lwd = c(2, 1.5))
```

## ৪.৫ পেয়ার্স প্লট (মাল্টিভেরিয়েবল অ্যানালিসিস)

## পর্ব ৫: মাল্টিপল প্লট (par() ফাংশন)

## মাস্টার পাইপলাইন: প্রোফেশনাল প্লট

```r
# সিস্টেমেটিকভাবে একটি সম্পূর্ণ প্লট তৈরি

# ১. ডাটা প্রস্তুত
data_to_plot <- mtcars
data_to_plot$cyl_factor <- as.factor(data_to_plot$cyl)

# ২. গ্রাফ প্যারামিটার সেট
par(mar = c(5, 5, 4, 2))  # মার্জিন সেট

# ৩. বেস প্লট
plot(data_to_plot$wt, data_to_plot$mpg,
     main = "গাড়ির কর্মদক্ষতা বিশ্লেষণ",
     xlab = "ওজন (1000 পাউন্ড)",
     ylab = "মাইল প্রতি গ্যালন\n(MPG)",
     pch = c(16, 17, 18)[as.numeric(data_to_plot$cyl_factor)],  # সিলিন্ডার অনুযায়ী পয়েন্ট
     col = c("#FF6B6B", "#4ECDC4", "#45B7D1")[as.numeric(data_to_plot$cyl_factor)],
     cex = 1.5,
     las = 1)  # অ্যাক্সিস লেবেল orientation

# ৪. ট্রেন্ড লাইন
abline(lm(mpg ~ wt, data = data_to_plot),
       col = "black",
       lwd = 2,
       lty = 2)

# ৫. গ্রিড
grid(col = "gray80", lty = "dotted")

# ৬. লেজেন্ড
legend("topright",
       legend = levels(data_to_plot$cyl_factor),
       title = "সিলিন্ডার সংখ্যা",
       pch = c(16, 17, 18),
       col = c("#FF6B6B", "#4ECDC4", "#45B7D1"),
       pt.cex = 1.5,
       bg = "white",
       box.lty = 1)

# ৭. টেক্সট বক্স
text(x = max(data_to_plot$wt) * 0.9,
     y = max(data_to_plot$mpg) * 0.9,
     labels = paste("Total Cars:", nrow(data_to_plot)),
     col = "darkgray",
     font = 3)

# ৮. বক্স আঁকা
box(lwd = 2)
```
