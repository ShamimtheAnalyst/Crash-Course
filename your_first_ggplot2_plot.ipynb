### __আপনাকে যা জানতে হবে:__

#### __Grammar of Graphics__
__Grammar of Graphics__ হলো ডেটা ভিজ্যুয়ালাইজেশনের একটি তাত্ত্বিক কাঠামো, যা লেল্যান্ড উইলকিনসন (Leland Wilkinson) প্রথম প্রবর্তন করেন এবং পরবর্তীতে হ্যাডলি উইকহ্যাম (Hadley Wickham) এটি ggplot2 প্যাকেজের মাধ্যমে জনপ্রিয় করেন।

সহজ ভাষায় বলতে গেলে, একটি বাক্য যেমন সাবজেক্ট, ভার্ব এবং অবজেক্ট নিয়ে গঠিত হয়, একটি গ্রাফও ঠিক তেমনি কিছু নির্দিষ্ট "ব্যাকরণ" বা লেয়ার নিয়ে গঠিত হয়। এর মূল কথা হলো—একটি জটিল গ্রাফকে ছোট ছোট স্বাধীন লেয়ারে ভাগ করা।

নিচে গ্রামার অফ গ্রাফিক্সের প্রধান ৭টি লেয়ার বা উপাদান নিয়ে বিস্তারিত আলোচনা করা হলো:

__১. ডেটা (Data):__
এটি গ্রাফের সবচেয়ে নিচের লেয়ার। আপনার কাছে একটি পরিষ্কার ডেটাসেট (সাধারণত Data Frame বা Tibble) থাকতে হবে।

    নিয়ম: ggplot2 সবসময় টিডি (Tidy) ডেটা পছন্দ করে, যেখানে প্রতিটি কলাম একটি ভ্যারিয়েবল এবং প্রতিটি রো একটি অবজারভেশন।

__২. এসথেটিক ম্যাপিং (Aesthetics - aes):__
এই লেয়ারে আমরা ঠিক করি ডেটার কোন ভ্যারিয়েবল গ্রাফের কোন দৃশ্যমান বৈশিষ্ট্যের সাথে যুক্ত হবে।

    উদাহরণ: x-অক্ষ বরাবর displ এবং y-অক্ষ বরাবর hwy বসানো। এছাড়া রঙ (color), আকার (size), এবং আকৃতি (shape) এই লেয়ারের অংশ।

__৩. জিওমেট্রি (Geometries - geom):__
এটি নির্ধারণ করে ডেটাটি গ্রাফে কীভাবে ফুটে উঠবে। একই ডেটা এবং এসথেটিক্স ব্যবহার করে আপনি ভিন্ন ভিন্ন জিওমেট্রি তৈরি করতে পারেন।

- geom_point(): স্ক্যাটার প্লটের জন্য।
- geom_bar(): বার চার্টের জন্য।
- geom_line(): লাইন চার্টের জন্য।

__৪. ফ্যাসেটস (Facets):__
ফ্যাসেটিংয়ের মাধ্যমে আমরা বড় ডেটাসেটকে ছোট ছোট ভাগে ভাগ করে একাধিক সাব-প্লট তৈরি করি। এটি ডেটার ভেতরের সাব-গ্রুপগুলোর তুলনা করতে সাহায্য করে।

    উদাহরণ: facet_wrap(~ class) দিলে প্রতিটি গাড়ির টাইপের জন্য আলাদা আলাদা ছোট গ্রাফ তৈরি হবে।

__৫. স্ট্যাটিস্টিকস (Statistics - stat):__
অনেক সময় আমরা সরাসরি ডেটা না দেখিয়ে তার ওপর কোনো গাণিতিক হিসাব দেখাতে চাই।

    উদাহরণ: একটি হিস্টোগ্রামে ডেটার বিন্যাস দেখানো বা geom_smooth() ব্যবহার করে একটি রিগ্রেশন লাইন (Regression Line) টানা।

__৬. কোঅর্ডিনেট সিস্টেম (Coordinates - coord):__
এটি গ্রাফের দ্বিমাত্রিক স্থান নির্ধারণ করে। সাধারণত আমরা Cartesian coordinate (X, Y) ব্যবহার করি।

    উদাহরণ: coord_flip() ব্যবহার করে x এবং y অক্ষ অদলবদল করা যায়, অথবা coord_polar() দিয়ে পাই চার্ট তৈরি করা যায়।

__৭. থিম (Themes):__
এটি গ্রাফের অ-ডেটা সংক্রান্ত বিষয়গুলো নিয়ন্ত্রণ করে (যেমন: ফন্ট সাইজ, ব্যাকগ্রাউন্ড কালার, লিজেন্ড পজিশন)। এটি গ্রাফের তথ্য পরিবর্তন করে না, শুধু সৌন্দর্য বাড়ায়।

#### __পাইপ অপারেটর কী এবং কেন?__

সাধারণ R কোডে আমাদের অনেকগুলো ফাংশন একসাথে ব্যবহার করতে হলে একটির ভেতর আরেকটি (Nested) লিখতে হয়, যা পড়া খুব কঠিন। পাইপ অপারেটর এই সমস্যা সমাধান করে ডেটাকে বাম দিক থেকে ডান দিকে প্রবাহিত করে।

উদাহরণ: ধরুন আপনি সকালে ঘুম থেকে উঠে চা বানাবেন।

- সাধারণ পদ্ধতি (Nested): চিনি(দুধ(গরম_পানি(চা_পাতা(কেটলি))))

- পাইপ পদ্ধতি: কেটলি %>% চা_পাতা %>% গরম_পানি %>% দুধ %>% চিনি!

### ৩. বার চার্ট তৈরি (Grammar of Graphics Layers)

#### লেয়ার ১ ও ২: ডেটা এবং এসথেটিক্স (Mapping)
প্রথমে আমরা ঠিক করব কোন ডেটা ব্যবহার করব এবং X-অক্ষ বরাবর কী থাকবে। এখানে আমরা প্রতিটি প্রজাতির ফুলের সংখ্যা গুনতে চাই, তাই x = Species দেব।

```r
# শুধুমাত্র ক্যানভাস তৈরি হবে
iris %>%
  ggplot(aes(x = Species))
```

#### লেয়ার ৩: জিওমেট্রি (Geometry)
এখন আমরা বার চার্টের আকৃতি দেওয়ার জন্য geom_bar() ফাংশনটি যোগ করব।

```r
# বার চার্ট দৃশ্যমান হবে
iris %>%
  ggplot(aes(x = Species)) +
  geom_bar()
```

#### লেয়ার ৪: কালার ও ফিল (Aesthetics)
গ্রাফটিকে সুন্দর করতে আমরা প্রতিটি প্রজাতির বার-এ আলাদা রঙ যোগ করব।

```r
# প্রজাতি অনুযায়ী রঙ (color) যোগ করা (outline color)
iris %>%
  ggplot(aes(x = Species, color = Species)) +
  geom_bar()
```

```r
# প্রজাতি অনুযায়ী রঙ (Fill) যোগ করা
iris %>%
  ggplot(aes(x = Species, fill = Species)) +
  geom_bar()
```

```r
# প্রজাতি অনুযায়ী রঙ (Fill) যোগ করা
iris %>%
  ggplot(aes(x = Species, fill = Species)) +
  geom_bar(alpha = 0.8)    # বারের পুরুত্ব
```

```r
# প্রজাতি অনুযায়ী রঙ (Fill) যোগ করা
iris %>%
  ggplot(aes(x = Species, fill = Species)) +
  geom_bar(alpha = 0.8,
          color = "black")    # বারের পুরুত্ব
```

```r
# প্রজাতি অনুযায়ী রঙ (Fill) যোগ করা
iris %>%
  ggplot(aes(x = Species, fill = Species)) +
  geom_bar(alpha = 0.8,
          color = "black",
          size = 0.8)    # বারের পুরুত্ব
```

```r
# প্রজাতি অনুযায়ী রঙ (Fill) যোগ করা
iris %>%
  ggplot(aes(x = Species, fill = Species)) +
  geom_bar(alpha = 0.8,
          color = "black",
          size = 0.8,
      width = 0.5)    # বারের পুরুত্ব
```

#### লেয়ার ৫: ফ্যাসেট (Faceting)
জটিল ডেটার ক্ষেত্রে আমরা ফ্যাসেট ব্যবহার করি। যদিও iris ডেটাসেটে এটি প্রয়োজন নেই, তবুও শেখার খাতিরে আমরা প্রতিটি প্রজাতির জন্য আলাদা সাব-প্লট তৈরি করতে পারি।

```r
# প্রতিটি প্রজাতির জন্য আলাদা ছোট গ্রাফ
iris %>%
  ggplot(aes(x = Species, fill = Species)) +
  geom_bar() +
  facet_wrap(~ Species)
```

#### লেয়ার ৬ ও ৭: লেবেল ও থিম (Labels & Themes)
সবশেষে আমরা শিরোনাম (Title) এবং একটি পরিষ্কার ব্যাকগ্রাউন্ড (Theme) যোগ করে গ্রাফটিকে পাবলিকেশন কোয়ালিটির করে তুলব।

```r
# পূর্ণাঙ্গ গ্রাফ কোড
iris_plot <- iris %>%
  ggplot(aes(x = Species, fill = Species)) +
  geom_bar() +
  labs(
    title = "Bar Chart: for Iris Data set",
    subtitle = "This dataset includes 150 flower samples",
    x = "Species Name",
    y = "Numbers of Samples",
    fill = "Species Type"
  ) +
  theme_minimal() # ক্লিন ব্যাকগ্রাউন্ডের জন্য

# গ্রাফটি দেখানো
iris_plot
```

### ৪. ফাইল সেভ করা (Saving the Output)
আপনার তৈরি করা গ্রাফটি হাই-কোয়ালিটি ইমেজ ফাইল হিসেবে সেভ করার জন্য ggsave() ফাংশনটি ব্যবহার করুন

```r
# গ্রাফটি .png ফরমেটে সেভ করা
ggsave("iris_bar_chart.png",
       plot = iris_plot,
       width = 7,
       height = 5,
       units = 'cm',
       dpi = 300)
```
