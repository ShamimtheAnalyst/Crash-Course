## __Mastering Histograms: From Base R to Professional ggplot2__

### নোটবুক পরিচিতি (Metadata)

- Aim: এই গাইডের মূল উদ্দেশ্য হলো R-এর ডিফল্ট গ্রাফিক্স এবং ggplot2 ব্যবহার করে একটি সাধারণ হিস্টোগ্রামকে ধাপে ধাপে কীভাবে প্রফেশনাল এবং ইনফরমেটিভ করা যায় তা শেখা।
- Target: যারা একদম নতুন (Beginners) এবং ডেটা ভিজ্যুয়ালাইজেশনের লেয়ারগুলো বুঝতে চান।
- Dataset: বিল্ট-ইন iris ডাটাসেট।

### আমরা এই লেসন থেকে যা শিখব (Learning Objectives):

- ১. Base R vs ggplot2: দুটির মধ্যে সিনট্যাক্সের পার্থক্য।
- ২. Layered Graphics: কীভাবে একটির পর একটি লেয়ার যোগ করে গ্রাফ উন্নত করতে হয়।
- ৩. Aesthetics Control: রঙ (fill), স্বচ্ছতা (alpha), এবং বর্ডার (color) নিয়ন্ত্রণ।
- ৪. Data Transformation: গ্রাফ তৈরির আগেই ডেটা ফিল্টার করার পদ্ধতি।
- ৫. Theming: গ্রাফের ব্যাকগ্রাউন্ড এবং টাইটেল পজিশন ঠিক করা।

### ১. বেসিক হিস্টোগ্রাম (Base R Graphics)
R-এ কোডিং শেখার শুরুতে আমরা এই ডিফল্ট ফাংশনটি ব্যবহার করি। এটি দ্রুত কোনো ডেটার ডিস্ট্রিবিউশন দেখার জন্য খুব কার্যকর।

```r
# ডিফল্ট ফাংশন ব্যবহার করে হিস্টোগ্রাম
hist(iris$Sepal.Length, breaks = 20, col = "skyblue",
     xlab = "Sepal Length", ylab = "Frequency",
     main = "Histogram of Sepal Length")
```

__ব্যাখ্যা:__ এখানে breaks দিয়ে বারের সংখ্যা ঠিক করা হয়েছে, আর col দিয়ে পুরো বারে বেগুনি রঙ দেওয়া হয়েছে। এটি দ্রুত কাজ করলেও কাস্টমাইজেশনের সুযোগ এখানে কিছুটা কম।

### ২. আধুনিক ভিজ্যুয়ালাইজেশন: ggplot2 (Step-by-Step)
ডেটা সায়েন্সের জগতে ggplot2 হলো সবচেয়ে জনপ্রিয় লাইব্রেরি। এটি Grammar of Graphics মেনে চলে, অর্থাৎ এখানে প্রতিটি অংশ একটি আলাদা লেয়ার।

#### ধাপ ১: ভিত্তি তৈরি (Foundation)

```r
library(tidyverse)

iris %>%
  ggplot(aes(Sepal.Length)) +
  geom_histogram()
```

__ব্যাখ্যা:__ শুরুতে আমরা শুধু ডেটা এবং x অক্ষ সেট করেছি। geom_histogram() বলার পর এটি একটি ডিফল্ট কালো হিস্টোগ্রাম তৈরি করেছে।

### ধাপ ২-৩: রঙ এবং স্বচ্ছতা (Color & Alpha)

```r
# color
iris %>%
  ggplot(aes(Sepal.Length)) +
  geom_histogram(fill = "blue")
```

```r
# Color transperancy
iris %>%
  ggplot(aes(Sepal.Length)) +
  geom_histogram(fill = "blue", alpha = 0.4)
```

**ব্যাখ্যা:** fill বারগুলোর ভেতরের রঙ বদলায়। alpha ব্যবহার করা হয়েছে স্বচ্ছতা (Transparency) দেওয়ার জন্য, যাতে বারগুলো খুব বেশি গাঢ় না দেখায়।

### ধাপ ৪: বর্ডার বা আউটলাইন যোগ করা

```r
iris %>%
  ggplot(aes(Sepal.Length)) +
  geom_histogram(fill = "blue", alpha = 0.4, color = "black")
```

**ব্যাখ্যা:** এখানে color = "black" প্রতিটি বারের চারপাশে একটি কালো বর্ডার তৈরি করেছে, যা প্রতিটি বিন-কে আলাদাভাবে চিনতে সাহায্য করে।

### ধাপ ৫: বিন-উইডথ এবং লেবেলিং (The Logic of Bins)

```r
iris %>%
  ggplot(aes(Sepal.Length)) +
  geom_histogram(fill = "blue", alpha = 0.4, color = "black", binwidth = 0.2) +
  labs(title = "Histogram of Sepal Length", x = "Sepal Length", y = "Frequency")
```

**ব্যাখ্যা:** binwidth = 0.2 খুবই গুরুত্বপূর্ণ। এটি ঠিক করে প্রতিটি বার কতটুকু চওড়া হবে। labs() ফাংশন দিয়ে আমরা গ্রাফের শিরোনাম এবং অক্ষের নামগুলো সুন্দর করে সাজিয়েছি।

### ধাপ ৬-৭: থিম এবং ডেটা ফিল্টারিং

```r
iris %>%
  filter(Sepal.Length > 5) %>% # ৫-এর চেয়ে বড় লেংথ ফিল্টার করা
  ggplot(aes(Sepal.Length)) +
  geom_histogram(fill = "blue", alpha = 0.4, color = "black", binwidth = 0.2) +
  theme_test() # ক্লিন ব্যাকগ্রাউন্ড
```

**ব্যাখ্যা:** filter() ব্যবহার করে আমরা অপ্রয়োজনীয় ছোট ডেটা বাদ দিয়েছি। theme_test() গ্রাফের ব্যাকগ্রাউন্ড থেকে অপ্রয়োজনীয় গ্রিড লাইন সরিয়ে সেটিকে আরও পরিষ্কার (Clean) করেছে।

### ধাপ ৮: চূড়ান্ত টাচ (Centering the Title)

```r
iris %>%
  filter(Sepal.Length > 5) %>%
  ggplot(aes(Sepal.Length)) +
  geom_histogram(fill = "blue", alpha = 0.4, color = "black", binwidth = 0.2) +
  labs(title = "Histogram of Sepal Length", x = "Sepal Length", y = "Frequency") +
  theme_test() +
  theme(plot.title = element_text(hjust = 0.5, face = "bold")) # টাইটেল মাঝখানে আনা
```

**ব্যাখ্যা:** ডিফল্টভাবে টাইটেল বাম দিকে থাকে। theme(plot.title = element_text(hjust = 0.5)) কোডটি শিরোনামকে ঠিক মাঝখানে (Center align) নিয়ে আসে।

#### __এক নজরে আপনার শেখা সূত্রগুলো (Cheat Sheet):__

- fill: বারের ভেতরের রঙ।
- color: বারের বর্ডার বা লাইনের রঙ।
- alpha: স্বচ্ছতা (০ থেকে ১-এর মধ্যে)।
- binwidth: প্রতিটি বারের ব্যাপ্তি বা উইডথ।
- hjust = 0.5: টাইটেল বা টেক্সটকে মাঝখানে আনা।
