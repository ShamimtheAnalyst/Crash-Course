### ২. এক নজরে ডিস্ট্রিবিউশন: Histograms

__উদ্দেশ্য:__ একটি কন্টিনিউয়াস ভ্যারিয়েবল (যেমন দৈর্ঘ্য) ডেটাসেটে কীভাবে বিন্যস্ত আছে তা দেখতে ব্যবহার করা হয়।

```r
iris %>%
  ggplot(aes(Sepal.Length)) +
  geom_histogram(fill = "steelblue", color = "white", bins = 20) +
  theme_test() +
  labs(title = "Sepal Length Distribution", x = "Length", y = "Frequency")
```

__ব্যাখ্যা:__ geom_histogram() ডেটাকে বিভিন্ন 'bins' বা গ্রুপে ভাগ করে ফ্রিকোয়েন্সি দেখায়। এটি ডেটা কি নরমাল ডিস্ট্রিবিউশনে আছে নাকি একদিকে ঝুঁকে আছে তা বুঝতে সাহায্য করে।

### ৩. ক্যাটাগরি তুলনা: Bar Chart

__উদ্দেশ্য:__ বিভিন্ন ক্যাটাগরির (যেমন বৈবাহিক অবস্থা) সংখ্যা তুলনা করতে এটি সেরা।

```r
gss_cat %>%
  ggplot(aes(marital)) +
  geom_bar(fill = "skyblue", color = "black") +
  theme_test() +
  labs(title = "Marital Status Count", x = "Status", y = "Count")
```

__ব্যাখ্যা:__ geom_bar() স্বয়ংক্রিয়ভাবে প্রতিটি ক্যাটাগরির সংখ্যা (count) গণনা করে এবং স্তম্ভ আকারে দেখায়।

### ৪. ডেটা সামারি ও আউটলায়ার: Box Plots

__উদ্দেশ্য:__ ডেটার গড় (Median), রেঞ্জ এবং আউটলায়ার (অস্বাভাবিক মান) শনাক্ত করতে ব্যবহৃত হয়।


```r
chickwts %>%
  ggplot(aes(weight, feed, fill = feed)) +
  geom_boxplot(alpha = 0.6) +
  theme_test(base_size = 15) +
  labs(title = "Chicken Weight by Feed Type", x = "Weight", y = "Feed Type")
```

__ব্যাখ্যা:__ মাঝখানের দাগটি হলো Median। বক্সের বাইরের বিন্দুগুলো হলো Outliers। এটি বিভিন্ন ফিড টাইপের কার্যকারিতা তুলনা করতে সাহায্য করছে।

### ৫. সময়ের ধারা: Time-series

__উদ্দেশ্য:__ সময়ের সাথে কোনো পরিবর্তন (যেমন সঞ্চয় হার) পর্যবেক্ষণ করতে ব্যবহৃত হয়।

```r
economics %>%
  drop_na() %>%
  ggplot(aes(date, psavert)) +
  geom_line(color = "steelblue", size = 1) +
  theme_test(base_size = 15) +
  labs(title = "Personal Savings Rate Over Time", x = "Year", y = "Savings Rate (%)")
```

### ৬. সম্পর্ক নির্ণয়: Scatter Plots & Regression

__উদ্দেশ্য:__ দুটি ভ্যারিয়েবলের মধ্যে সম্পর্ক (Correlation) এবং ট্রেন্ড লাইন দেখতে ব্যবহৃত হয়।

```r
penguins %>%
  drop_na(body_mass_g, flipper_length_mm) %>%
  ggplot(aes(flipper_length_mm, body_mass_g)) +
  geom_point(aes(color = species), alpha = 0.7, size = 3) +
  geom_smooth(method = "lm", color = "red", se = TRUE) + # Regression Line
  theme_test(base_size = 15) +
  labs(title = "Flipper Length vs Body Mass", subtitle = "Linear Regression by Species")
```

```r
penguins %>%
  drop_na(body_mass_g, flipper_length_mm) %>%
  ggplot(aes(flipper_length_mm, body_mass_g)) +
  geom_point(aes(color = species), alpha = 0.7, size = 3) +
  geom_smooth(method = "lm", color = "red", se = TRUE) + # Regression Line
  facet_wrap(~species) + # প্রজাতি অনুযায়ী আলাদা গ্রাফ
  theme_test(base_size = 15) +
  labs(title = "Flipper Length vs Body Mass", subtitle = "Linear Regression by Species")
```

__ব্যাখ্যা:__ geom_point() বিন্দুগুলো দেখায় এবং geom_smooth(method = "lm") একটি লিনিয়ার রিগ্রেশন লাইন যোগ করে, যা আমাদের জানায় একটি ভ্যারিয়েবল বাড়লে অন্যটি বাড়ার হার কেমন।

### ৭. প্রফেশনাল কাস্টমাইজেশন (Advanced Plot)

__উদ্দেশ্য:__ রিপোর্ট বা প্রেজেন্টেশনের জন্য গ্রাফকে সুন্দর করে সাজানো।

```r
Salaries %>%
  filter(salary < 220000) %>%
  ggplot(aes(rank, salary, fill = sex)) +
  geom_boxplot(alpha = 0.5) +
  theme_test(base_size = 15) +
  scale_y_continuous(labels = c("$50k", "$100k", "$150k", "$200k"),
                     breaks = c(50000, 100000, 150000, 200000)) +
  scale_x_discrete(labels = c("AsstProf" = "Assistant\nProfessor",
                              "AssocProf" = "Associate\nProfessor",
                              "Prof" = "Professor")) +
  theme(legend.position = "top", axis.text.x = element_text(angle = 45, hjust = 1)) +
  labs(title = "Faculty Salary Analysis", fill = "Gender")
```

__ব্যাখ্যা:__ এখানে আমরা scale_y_continuous দিয়ে অক্ষের মান সুন্দর করেছি এবং theme() দিয়ে লিজেন্ডের অবস্থান ও ফন্ট ঠিক করেছি। এটি একটি 'Publication Ready' প্লট।

#### এক নজরে চেকলিস্ট:

- Histogram/Density: একটি সংখ্যাসূচক ভ্যারিয়েবলের জন্য।
- Bar Chart: ক্যাটাগরিক্যাল ভ্যারিয়েবলের সংখ্যা দেখতে।
- Scatter Plot: দুটি সংখ্যার সম্পর্ক দেখতে।
- Boxplot: ডিস্ট্রিবিউশন এবং আউটলায়ার দেখতে।
