## **Mastering Loops in R: From Basics to Flow Control**

## **১. লুপ কী? (What is a Loop?)**

লুপ হলো এমন একটি প্রোগ্রামিং স্ট্রাকচার যা একটি নির্দিষ্ট কোড ব্লক বারবার রান করে। ডেটা অ্যানালাইসিসে যখন আমাদের ডাটাসেটের প্রতিটি রো (row) বা আইটেমের ওপর একই কাজ করতে হয়, তখন লুপ ব্যবহার করা হয়। আর (R) ল্যাঙ্গুয়েজে সবচেয়ে জনপ্রিয় লুপ হলো `for loop`।

```r
#install.packages('tidyverse')
library(tidyverse)
```

## **২. বেসিক ফর লুপ (Basic For Loop)**

নিচে স্টার ওয়ার্স (Star Wars) ডাটাসেটের প্রথম ৫টি চরিত্রের নাম প্রিন্ট করার পদ্ধতি দেখানো হলো:

```r
# i হলো একটি কাউন্টার (1, 2, 3...)
# 1:5 মানে হলো "কোডটি ৫ বার রান করো"
for(i in 1:5) {
  print(starwars$name[i])
}
```

## **৩. ডায়নামিক লুপিং (Looping Through All Data)**

কোডে সংখ্যা (যেমন ৫ বা ১০) নির্দিষ্ট না করে `length()` ফাংশন ব্যবহার করা ভালো। এতে ডাটাসেটে যতগুলো ডেটা থাকবে, লুপটি ঠিক ততবারই চলবে।

```r
# পুরো ডাটাসেটের সব উচ্চতা প্রিন্ট করা
for(i in 1:length(starwars$height)) {
  print(starwars$height[i])
}
```

## **৪. লুপের ফলাফল স্টোর করা (Storing Loop Results)**

লুপের কাজ শুধু প্রিন্ট করা নয়, বরং ক্যালকুলেশন করে ফলাফলগুলো একটি ভেক্টরে সেভ করে রাখা। এখানে আমরা সেন্টিমিটার থেকে মিটার-এ কনভার্ট করব।

```r
# ১. একটি খালি ভেক্টর তৈরি করা (৫টি সংখ্যা রাখার জন্য)
tallness <- numeric(length = 5)

# ২. লুপের মাধ্যমে ডাটা ক্যালকুলেট করে স্টোর করা
for(i in 1:5) {
  tallness[i] <- starwars$height[i] / 100
}

# ফলাফল চেক করা
tallness
```

## **৫. লুপ কন্ট্রোল: break এবং next**

লুপ চলার সময় মাঝপথে সেটি থামানো বা কোনো অংশ বাদ দেওয়ার জন্য এই দুটি কমান্ড ব্যবহার করা হয়।

### **ক) break (পুরোপুরি থেমে যাওয়া)**

যদি কোনো নির্দিষ্ট শর্ত পূরণ হয় এবং আপনি চান লুপটি আর না চলুক, তবে `break` ব্যবহার করুন।

```r
for(name in starwars$name) {
  print(name)
  if(name == "Darth Vader") {
    break  # ডার্থ ভেডার পাওয়ার সাথে সাথে লুপ বন্ধ হয়ে যাবে
  }
}
```

### **খ) next (নির্দিষ্ট আইটেম স্কিপ করা)**

কোনো নির্দিষ্ট আইটেম বাদ দিয়ে পরের আইটেমে চলে যেতে চাইলে `next` ব্যবহার করুন।

```r
for(name in starwars$name) {
  if(name == "C-3PO") {
    next  # C-3PO আসলে সেটি স্কিপ করবে এবং পরের নামে চলে যাবে
  }
  print(name)
}
```

## **৬. cat() ফাংশন দিয়ে প্রফেশনাল আউটপুট**

`print()` এর চেয়ে `cat()` ব্যবহার করলে টেক্সট এবং ভ্যারিয়েবল মিশিয়ে খুব সুন্দর মেসেজ তৈরি করা যায়।

```r
for(i in 1:5) {
  cat("✨", starwars$name[i], "is",
      tallness[i], "meters tall.\n")
}
```

### **🎓 কি-টেকঅ্যাওয়ে (Key Takeaways)**

* **`for (i in x)`**: x এর প্রতিটি আইটেমের জন্য কোড রান করে।
* **`break`**: লুপ থেকে তাৎক্ষণিক বেরিয়ে আসার জন্য।
* **`next`**: বর্তমান ইটারেশন বাদ দিয়ে পরবর্তী ইটারেশনে যাওয়ার জন্য।
* **`numeric()` / `character()**`: লুপের ফলাফল জমানোর জন্য খালি জায়গা তৈরি করা।
* **`cat()`**: সুন্দর করে মেসেজ সাজিয়ে আউটপুট দেখানোর জন্য।
